the for loop starts a variable avg at 0, then for every value in temps, it adds temp/4 to avg. Since there are exactly four values, dividing each one by 4 before adding it up is the same as adding all four values and then dividing the total by 4 to get the average.


```python
temps = [110, 105, 101, 103]

avg = 0
for temp in temps: 
    avg+= temp/4
```
