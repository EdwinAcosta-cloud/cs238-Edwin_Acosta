```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
x = np.arange(0, 12, 0.01)
y = np.arange(0, 12, 0.01)

xi, yi = np.meshgrid(x, y)

#zi = np.sin(xi)*np.sin(yi)

zb = np.sin(xi)**2+np.sin(yi)**2
```

```python
#zi
zb
```

```python
fig, ax = plt.subplots(figsize=(4,4))
ax.contour(xi,yi,zb, colors = 'k', levels = np.arange(-1, 1.1, 0.2))
```

```python
fig, ax = plt.subplots(figsize=(4,4))
ax.contourf(xi,yi,zb,  cmap=plt.cm.Blues ,levels = np.arange(-1, 1.1, 0.2))
```

```python
fig, ax = plt.subplots(figsize=(4,4))
ax.imshow(zb, origin = 'lower', extent = [0, 12, 0, 13])
```

```python
fig = plt.figure(figsize = (5,5))
ax = fig.add_subplot(111, projection = '3d')
ax.plot_surface(xi, yi, zb, cmap = 'RdBu')
ax.view_init(elev = 45, azim = 135)
```
